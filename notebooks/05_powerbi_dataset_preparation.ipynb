```python
import warnings
warnings.filterwarnings("ignore")

from pathlib import Path
import numpy as np
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")

print("Environment ready.")
print("Power BI dataset preparation initialized.")
```

```python
input_path = Path("../data/processed/airbnb_powerbi.csv")
output_dir = Path("../data/processed/powerbi")

output_dir.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(input_path, parse_dates=["last_review"])

print("Dataset loaded successfully.")
print("Shape:", df.shape)

df.head()
```

```python
df_powerbi = df.copy()

# Business-friendly field for host behavior
df_powerbi["host_type"] = np.where(
    df_powerbi["calculated_host_listings_count"].fillna(1) > 1,
    "Multi-listing host",
    "Single-listing host"
)

# Business-friendly field for review status
df_powerbi["review_status"] = np.where(
    df_powerbi["has_reviews"] == 1,
    "Has reviews",
    "No reviews"
)

df_powerbi.head()
```

```python
dashboard_columns = [
    "id",
    "name",
    "host_id",
    "host_name",
    "neighbourhood_group",
    "neighbourhood",
    "latitude",
    "longitude",
    "room_type",
    "price",
    "currency",
    "price_eur",
    "minimum_nights",
    "number_of_reviews",
    "last_review",
    "reviews_per_month",
    "calculated_host_listings_count",
    "availability_365",
    "city",
    "has_reviews",
    "review_status",
    "host_type",
]

df_powerbi = df_powerbi[dashboard_columns]

print("Power BI dataset shape:", df_powerbi.shape)
df_powerbi.head()
```

```python
df_powerbi.info()
```

```python
missing_summary = (
    df_powerbi
    .isna()
    .sum()
    .reset_index()
)

missing_summary.columns = ["column", "missing_count"]
missing_summary["missing_pct"] = (
    missing_summary["missing_count"] / len(df_powerbi) * 100
).round(2)

missing_summary.sort_values("missing_pct", ascending=False)
```

```python
summary_by_city = (
    df_powerbi
    .groupby("city")
    .agg(
        total_listings=("id", "count"),
        total_hosts=("host_id", "nunique"),
        avg_price_eur=("price_eur", "mean"),
        median_price_eur=("price_eur", "median"),
        avg_reviews=("number_of_reviews", "mean"),
        avg_reviews_per_month=("reviews_per_month", "mean"),
        avg_availability=("availability_365", "mean"),
        listings_with_reviews=("has_reviews", "sum"),
        multi_listing_host_listings=(
            "host_type",
            lambda x: (x == "Multi-listing host").sum()
        )
    )
    .reset_index()
)

summary_by_city["pct_listings_with_reviews"] = (
    summary_by_city["listings_with_reviews"] /
    summary_by_city["total_listings"] * 100
).round(2)

summary_by_city["pct_multi_listing_host_supply"] = (
    summary_by_city["multi_listing_host_listings"] /
    summary_by_city["total_listings"] * 100
).round(2)

summary_by_city
```

```python
summary_by_neighbourhood = (
    df_powerbi
    .groupby(["city", "neighbourhood"])
    .agg(
        total_listings=("id", "count"),
        avg_price_eur=("price_eur", "mean"),
        median_price_eur=("price_eur", "median"),
        avg_reviews=("number_of_reviews", "mean"),
        avg_availability=("availability_365", "mean")
    )
    .reset_index()
    .sort_values(["city", "total_listings"], ascending=[True, False])
)

summary_by_neighbourhood.head()
```

```python
main_output_path = output_dir / "airbnb_powerbi.csv"
city_summary_output_path = output_dir / "airbnb_powerbi_summary_by_city.csv"
neighbourhood_summary_output_path = output_dir / "airbnb_powerbi_summary_by_neighbourhood.csv"

df_powerbi.to_csv(main_output_path, index=False)
summary_by_city.to_csv(city_summary_output_path, index=False)
summary_by_neighbourhood.to_csv(neighbourhood_summary_output_path, index=False)

print("Power BI datasets exported successfully:")
print(main_output_path)
print(city_summary_output_path)
print(neighbourhood_summary_output_path)
```
